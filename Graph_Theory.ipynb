```python
import networkx as nx

# Build the weighted graph from the route data
routes = [
    ("A", "B", 4),
    ("A", "C", 2),
    ("B", "C", 1),
    ("B", "D", 5),
    ("C", "D", 3),
    ("C", "E", 6),
    ("D", "E", 2),
    ("D", "F", 4),
    ("E", "F", 3),
]

G = nx.Graph()
G.add_weighted_edges_from(routes, weight="cost")

print("=" * 55)
print("PART 1: SHORTEST ROUTE FROM A TO F (DIJKSTRA)")
print("=" * 55)
print("Nodes:", list(G.nodes()))
print("Routes and costs:")
for u, v, d in G.edges(data=True):
    print(f"  {u} - {v} : cost = {d['cost']}")

# Run Dijkstra from A to F
path = nx.dijkstra_path(G, source="A", target="F", weight="cost")
total_cost = nx.dijkstra_path_length(G, source="A", target="F", weight="cost")

print("\nShortest route :", " -> ".join(path))
print("Total cost     :", total_cost)
```

```python
import networkx as nx

routes = [
    ("A", "B", 4),
    ("A", "C", 2),
    ("B", "C", 1),
    ("B", "D", 5),
    ("C", "D", 3),
    ("C", "E", 6),
    ("D", "E", 2),
    ("D", "F", 4),
    ("E", "F", 3),
]

G = nx.Graph()
G.add_weighted_edges_from(routes, weight="cost")

print("=" * 55)
print("PART 2: MINIMUM SPANNING TREE (KRUSKAL)")
print("=" * 55)

mst = nx.minimum_spanning_tree(G, weight="cost", algorithm="kruskal")

print("Routes kept in the MST:")
for u, v, d in sorted(mst.edges(data=True), key=lambda e: e[2]["cost"]):
    print(f"  {u} - {v} : cost = {d['cost']}")

print(f"\nNumber of routes kept : {mst.number_of_edges()}")
print(f"Total connection cost : {mst.size(weight='cost')}")
print("Is it a valid spanning tree?", nx.is_tree(mst))

print("\nRoutes dropped:")
for u, v, d in sorted(G.edges(data=True), key=lambda e: e[2]["cost"]):
    if not mst.has_edge(u, v):
        print(f"  {u} - {v} : cost = {d['cost']}")
```

```python
import networkx as nx

# Directed routes, going from the earlier letter to the later one
capacity_routes = [
    ("A", "B", 20),
    ("A", "C", 15),
    ("B", "C", 10),
    ("B", "D", 12),
    ("C", "D", 18),
    ("C", "E", 10),
    ("D", "E", 15),
    ("D", "F", 20),
    ("E", "F", 12),
]

D = nx.DiGraph()
D.add_weighted_edges_from(capacity_routes, weight="capacity")

print("=" * 60)
print("PART 3: MAXIMUM FLOW FROM A TO F (DIRECTED)")
print("=" * 60)
print("Directed routes and capacities:")
for u, v, d in D.edges(data=True):
    print(f"  {u} -> {v} : capacity = {d['capacity']}")

flow_value, flow_dict = nx.maximum_flow(D, "A", "F", capacity="capacity")

print(f"\nMaximum flow from A to F : {flow_value} deliveries")

print("\nFlow on each route:")
for u in flow_dict:
    for v, f in flow_dict[u].items():
        if f > 0:
            print(f"  {u} -> {v} : flow = {f:>2} / capacity = {D[u][v]['capacity']}")

cut_value, (S, T) = nx.minimum_cut(D, "A", "F", capacity="capacity")
print(f"\nMinimum cut value : {cut_value}")
print("Source side (S)   :", sorted(S))
print("Sink side (T)     :", sorted(T))
print("Bottleneck routes :", [(u, v) for u in S for v in D.successors(u) if v in T])
```
